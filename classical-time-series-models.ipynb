```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose

# Veri okuma
df = pd.read_csv('lip_0_x.csv')

# Zaman serisini çiz (trend ve mevsimsellik gözlemi)
plt.figure(figsize=(12, 6))
plt.plot(df['time'], df['0_x'])
plt.title('Zaman Serisi Grafiği')
plt.xlabel('Zaman (saniye)')
plt.ylabel('0_x Koordinatı')
plt.grid(True)
plt.show()

# Durağanlık testi
adf_result = adfuller(df['0_x'])
print(f'ADF Test p-değeri: {adf_result[1]}')
if adf_result[1] < 0.05:
    print("Seri durağan.")
else:
    print("Seri durağan değil.")

# Mevsimsellik ve trend analizi (seasonal_decompose)
decompose_result = seasonal_decompose(df['0_x'], period=50, model='additive')

decompose_result.plot()
plt.show()

# Otokorelasyon fonksiyonu
plt.figure(figsize=(12, 6))
plot_acf(df['0_x'], lags=30)
plt.title('Otokorelasyon (ACF)')
plt.grid(True)
plt.show()

# Kısmi otokorelasyon fonksiyonu
plt.figure(figsize=(12, 6))
plot_pacf(df['0_x'], lags=30)
plt.title('Kısmi Otokorelasyon (PACF)')
plt.grid(True)
plt.show()
```
