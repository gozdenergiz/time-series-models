```python
import pandas as pd

# Dosyayı yükleme
file_path = "data/lip_coordinates.csv"
df = pd.read_csv(file_path)

# İlk 5 sütunun isimlerini al
first_5_columns = df.columns[1:6].tolist()

# Bu sütunların y değerlerini al
y_columns = [col.replace('_x', '_y') for col in first_5_columns if col.replace('_x', '_y') in df.columns]

# Yeni dataframe oluşturma
new_df_with_time = df[['time'] + first_5_columns + y_columns]

# Yeni CSV dosyasına kaydetme
new_file_path_with_time = "data/selected_lip_coordinates.csv"
new_df_with_time.to_csv(new_file_path_with_time, index=False)

# Kullanıcıya dosya bağlantısı sağlama
new_file_path_with_time
```

```python
import pandas as pd

# Dosyayı yükleme
file_path = "data/lip_coordinates.csv"
df = pd.read_csv(file_path)


# İlk 5 sütunun isimlerini al (sadece _x olanları seç)
first_5_columns = [col for col in df.columns if col.endswith('_x')][:5]

# Sadece "time" sütunu ve ilk 5 sütundaki x değerlerini seçme
new_df_x_only = df[['time'] + first_5_columns]

# Yeni CSV dosyasına kaydetme
new_file_path_x_only = "data/selected_lip_coordinates_x_only.csv"
new_df_x_only.to_csv(new_file_path_x_only, index=False)

print(f"Dosya başarıyla kaydedildi: {new_file_path_x_only}")
```

```python
import pandas as pd

# Dosyayı yükleme
file_path = "data/lip_coordinates.csv"
df = pd.read_csv(file_path)


# İlk 5 sütunun isimlerini al (sadece _x olanları seç)
first_5_x_columns = [col for col in df.columns if col.endswith('_x')][:5]

# Bu sütunların y değerlerini al (_x yerine _y olanları seç)
first_5_y_columns = [col.replace('_x', '_y') for col in first_5_x_columns if col.replace('_x', '_y') in df.columns]

# Sadece "time" sütunu ve ilk 5 sütunun y değerlerini seçme
new_df_y_only = df[['time'] + first_5_y_columns]

# Yeni CSV dosyasına kaydetme
new_file_path_y_only = "data/selected_lip_coordinates_y_only.csv"
new_df_y_only.to_csv(new_file_path_y_only, index=False)

print(f"Dosya başarıyla kaydedildi: {new_file_path_y_only}")
```
